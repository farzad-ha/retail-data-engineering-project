```sql
CREATE SCHEMA IF NOT EXISTS second_data_engineering_project.star_schema;
```

```sql
CREATE OR REPLACE VIEW second_data_engineering_project.star_schema.fact_orders_table
AS
SELECT
    o.order_id,
    oi.order_item_id,
    dc.customer_key,
    dp.product_key,
    ds.seller_key,
    dpay.payment_key,
    dr.review_key,
    o.order_status,
    o.order_purchase_timestamp,
    o.order_approved_at,
    o.order_delivered_carrier_date,
    o.order_delivered_customer_date,
    o.order_estimated_delivery_date,
    oi.shipping_limit_date,
    oi.price,
    oi.freight_value
FROM second_data_engineering_project.silver.orders AS o
INNER JOIN second_data_engineering_project.silver.order_items AS oi
ON o.order_id = oi.order_id
LEFT JOIN second_data_engineering_project.star_schema.dim_customers AS dc
    ON o.customer_id = dc.customer_id

LEFT JOIN second_data_engineering_project.star_schema.dim_products AS dp
    ON oi.product_id = dp.product_id

LEFT JOIN second_data_engineering_project.star_schema.dim_sellers AS ds
    ON oi.seller_id = ds.seller_id

LEFT JOIN second_data_engineering_project.star_schema.dim_payments AS dpay
    ON o.order_id = dpay.order_id

LEFT JOIN second_data_engineering_project.star_schema.dim_reviews AS dr
    ON o.order_id = dr.order_id;
```
