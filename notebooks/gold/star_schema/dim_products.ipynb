```sql
CREATE OR REPLACE VIEW second_data_engineering_project.star_schema.dim_products

SELECT
    ROW_NUMBER() OVER (ORDER BY product_id) AS product_key,
    product_id,
    product_category_name,
    product_name_length,
    product_description_length,
    product_photos_qty,
    product_weight_g,
    product_length_cm,
    product_height_cm,
    product_width_cm
FROM second_data_engineering_project.silver.products;
```
