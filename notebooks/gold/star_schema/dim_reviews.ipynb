```sql
CREATE OR REPLACE VIEW second_data_engineering_project.star_schema.dim_reviews
AS
SELECT
    ROW_NUMBER() OVER (ORDER BY review_id) AS review_key,
    review_id,
    order_id,
    review_score,
    review_comment_message,
    review_creation_date,
    review_answer_timestamp
FROM
    second_data_engineering_project.silver.reviews;
```
