```sql
CREATE OR REPLACE VIEW second_data_engineering_project.star_schema.dim_customers
AS

SELECT
    ROW_NUMBER() OVER (ORDER BY customer_id) AS customer_key,
    customer_id,
    customer_unique_id,
    customer_zip_code_prefix,
    customer_city,
    customer_state
FROM second_data_engineering_project.silver.customers;
```
