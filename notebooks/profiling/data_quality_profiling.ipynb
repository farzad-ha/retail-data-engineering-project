```python
from pyspark.sql import functions as F

df_customers = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/second_data_engineering_project/landing/raw_files/olist_customers_dataset.csv")
```

```python
df_customers.printSchema()

df_customers.show(10, truncate=False)

df_customers.count()
```

```python
df_customers.filter(
    ~F.col("customer_id").rlike("^[0-9a-fA-F]{32}$")
).count()
```

```python
df_customers.select("customer_id").distinct().count()
```

```python
df_customers.select("customer_unique_id").distinct().count()
```

```python
df_customers.select("customer_unique_id").distinct().count()
df_customers.filter(F.trim(F.col("customer_id")) == "").count()
df_customers.select("customer_id").distinct().count()
```

```python
df_customers.filter(F.trim(F.col("customer_id")) == "").count()
```

```python
df_customers.groupBy("customer_unique_id") \
    .count() \
    .filter(F.col("count") > 1) \
    .count()
```

```python
df_customers.groupBy("customer_unique_id") \
    .agg(F.count("*").alias("customer_count")) \
    .filter(F.col("customer_count") > 1) \
    .groupBy("customer_count") \
    .agg(F.count("*").alias("number_of_customers")) \
    .orderBy("customer_count") \
    .show()
```

```python
df_customers.select([
    F.sum(F.col(column_name).isNull().cast("int")).alias(column_name)

    for column_name in df_customers.columns
]).show()
```
